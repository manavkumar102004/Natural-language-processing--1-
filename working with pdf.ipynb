## Question 1: Working with PDF
**(a)** Read the PDF, print the number of pages and the text of every page.

**(b)** Search for the word *dashboard* and print the page numbers on which it is found.

```python
# Q1 (a): Number of pages and text of every page
from PyPDF2 import PdfReader

reader=PdfReader('powerbi_visualization.pdf')
print('Number of pages: ',len(reader.pages))

for page in reader.pages:
  print(page.extract_text())
```

```python
# Q1 (b): Search a word in the PDF
search_word='dashboard'

for page_number,page in enumerate(reader.pages,start=1):
  if search_word in page.extract_text():
    print('Found on page',page_number)
```

## Question 2: Extract sentences from the PDF
**(a)** Extract the text of all the pages of the PDF into one text.

**(b)** Split the text into sentences with spaCy, print the number of sentences and the first 5 sentences.

```python
# Q2 (a): Text of all the pages
text=''
for page in reader.pages:
  text+=page.extract_text()

print(text[:300])
```

```python
# Q2 (b): Split the text into sentences
import spacy
nlp=spacy.load('en_core_web_sm')

doc=nlp(text)
sentences=[sentence.text for sentence in doc.sents]    # list of all sentences

print('Number of sentences:',len(sentences))
for sentence in sentences[:5]:
  print(sentence)
```

## Question 3: Tokenization
**(a)** Tokenize the first sentence of the PDF and print the index, the token and its attributes (alpha, punct, space).

**(b)** Tokenize the second sentence and remove the punctuation.

```python
# Q3 (a): Tokens with index and attributes
doc=nlp(sentences[0])
for token in doc:
  print(token.i,
        token.text,
        token.is_alpha,
        token.is_punct,
        token.is_space)
```

```python
# Q3 (b): Remove punctuation after tokenization
doc=nlp(sentences[1])
for token in doc:
  if not token.is_punct:
    print(token.text)
```

## Question 4: Stop words removal
**(a)** Find the stop words in the first sentence of the PDF.

**(b)** Remove the stop words and punctuation from the first three sentences using a function.

```python
# Q4 (a): Find the stop words
doc=nlp(sentences[0])
stopwords=[]

for token in doc:
  if token.is_stop:
    stopwords.append(token.text)

print(stopwords)
```

```python
# Q4 (b): Remove stop words and punctuation
def remove_stopwords(text):
    doc = nlp(text)
    result = []

    for token in doc:
        if not token.is_stop and not token.is_punct:
            result.append(token.text)

    return " ".join(result)

for sentence in sentences[:3]:
  print(remove_stopwords(sentence))
```

## Question 5: Lemmatization
**(a)** Print the lemma of every token of the first sentence of the PDF.

**(b)** Lemmatize the first three sentences after removing the stop words and punctuation using a function.

```python
# Q5 (a): Token and its lemma
doc=nlp(sentences[0])
for token in doc:
  print(token.text, "->", token.lemma_)
```

```python
# Q5 (b): Lemmatize after removing stop words and punctuation
def lemmatize_text(text):
  doc=nlp(text)
  result=[]
  for token in doc:
    if not token.is_stop and not token.is_punct:
      result.append(token.lemma_)

  return " ".join(result)

for sentence in sentences[:3]:
  print(lemmatize_text(sentence))
```

## Question 6: Stemming
**(a)** Stem every word of the first sentence of the PDF using NLTK's `PorterStemmer`.

**(b)** Print the original text and the stemmed text of the third sentence.

```python
# Q6 (a): Stem every word
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()

doc=nlp(sentences[0])
for token in doc:
  if token.is_alpha:
    print(token.text,"->",stemmer.stem(token.text))
```

```python
# Q6 (b): Original text and stemmed text
doc=nlp(sentences[2])

stemmed_words=[]
for token in doc:
  if token.is_alpha:
    stemmed_words.append(stemmer.stem(token.text.lower()))

stemmed_text=" ".join(stemmed_words)
print('Original:')
print(sentences[2])

print("\nStemmed:")
print(stemmed_text)
```

```python
# Q8 (a): All entities with label and meaning
doc=nlp(text)
for ent in doc.ents:
  print('Entity:',ent.text,
        '|Label:',ent.label_,
        '|Meaning:',spacy.explain(ent.label_))
```

## Question 9: Vocabulary and rule-based matching (Matcher)
**(a)** Create a `Matcher` and add patterns for *Power BI* (3 patterns) and *drill down* (2 patterns).

**(b)** Match the patterns against the PDF text and print the matches.

```python
# Q9 (a): Create the matcher and add patterns
from spacy.matcher import Matcher
matcher=Matcher(nlp.vocab)

# Power BI -> powerbi / power bi / power-bi
pattern1=[{'LOWER':'powerbi'}]
pattern2=[{'LOWER':'power'},{'LOWER':'bi'}]
pattern3=[{'LOWER':'power'},{'IS_PUNCT':True},{'LOWER':'bi'}]
matcher.add('PowerBI',[pattern1,pattern2,pattern3])

# drill down -> drill down / drill-down
pattern4=[{'LOWER':'drill'},{'LOWER':'down'}]
pattern5=[{'LOWER':'drill'},{'IS_PUNCT':True},{'LOWER':'down'}]
matcher.add('DrillDown',[pattern4,pattern5])
```

```python
# Q9 (b): Find and print the matches
doc=nlp(text)
found_matches=matcher(doc)

for match_id,start,end in found_matches:
  string_id=nlp.vocab.strings[match_id]
  span=doc[start:end]
  print(match_id,string_id,start,end,span.text)
```

## Question 10: PhraseMatcher
**(a)** Create a `PhraseMatcher` and add the phrases *Power Query*, *Looker Studio*, *Google Analytics* and *SQL Server*.

**(b)** Match the phrases against the PDF text and print the matches.

```python
# Q10 (a): Create the phrase matcher and add the phrases
from spacy.matcher import PhraseMatcher
matcher=PhraseMatcher(nlp.vocab)

phrase_list=['Power Query','Looker Studio','Google Analytics','SQL Server']
phrase_patterns=[nlp(text) for text in phrase_list]

matcher.add('PDFPhrases',phrase_patterns)
```

```python
# Q10 (b): Find and print the matches
doc=nlp(text)
matches=matcher(doc)
print(matches)

for match_id,start,end in matches:
  print(doc[start:end].text)
```
