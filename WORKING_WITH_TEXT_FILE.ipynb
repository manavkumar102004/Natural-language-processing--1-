## Question 7

Read `powerbi_visualization.txt`, tokenize it using NLTK and count how many stop words it contains.

```python
# Q7: Stop words count in powerbi_visualization.txt using nltk
import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')
stop_words=set(stopwords.words('english'))

with open('powerbi_visualization.txt','r',encoding='utf-8') as file:
  text=file.read()

tokens=word_tokenize(text)
count=0
for word in tokens:
  if word.lower() in stop_words:
    count+=1

print('Number of stopwords:',count)
```

## Question 8

Read `powerbi_visualization.txt` and count how many words of each part of speech (POS) it has. Print the counts.

```python
# Q8: POS tag count of powerbi_visualization.txt using spacy
from collections import Counter

with open('powerbi_visualization.txt','r',encoding='utf-8') as file:
  text=file.read()

doc=nlp(text)

pos_counts=Counter(token.pos_ for token in doc)
print(pos_counts)
```

## Question 9

Read `powerbi_visualization.txt`, take its words in lowercase and print the first 20 words with their stem using the Porter stemmer.

```python
# Q9: Stemming of powerbi_visualization.txt using nltk PorterStemmer
import re
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()

with open('powerbi_visualization.txt','r',encoding='utf-8') as file:
  text=file.read()

words=re.findall(r'\b[a-zA-Z]+\b',text.lower())

for word in words[:20]:
  print(word, "->", stemmer.stem(word))
```

## Question 10

Read `powerbi_visualization.txt` and print the lemmas of the first 30 words that are not stop words or punctuation, using spaCy.

```python
# Q10: Lemmatization of powerbi_visualization.txt using spacy
with open('powerbi_visualization.txt','r',encoding='utf-8') as file:
  text=file.read()

doc=nlp(text)
words=[]
for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.lemma_)

print(words[:30])
```

## Question 11

Use spaCy's `PhraseMatcher` to find the phrases `bar chart`, `line chart`, `pie chart` and `donut chart` in `powerbi_visualization.txt`. Print each matched phrase.

```python
# Q11: Phrase matching in powerbi_visualization.txt using PhraseMatcher
from spacy.matcher import PhraseMatcher
matcher=PhraseMatcher(nlp.vocab)

with open('powerbi_visualization.txt', 'r', encoding='utf-8') as f:
    doc3 = nlp(f.read())

phrase_list=['bar chart','line chart','pie chart','donut chart']

phrase_patterns=[nlp(text) for text in phrase_list]

matcher.add('ChartTypes',phrase_patterns)

matches=matcher(doc3)

for match_id,start,end in matches:
  string_id=nlp.vocab.strings[match_id]
  span=doc3[start:end]
  print(string_id,start,end,span.text)
```

Read `powerbi_visualization.txt` and use spaCy NER to print the first 10 named entities with their labels.
